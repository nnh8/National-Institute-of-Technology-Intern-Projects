```python
import PyPDF2

o_lines=[]
i=500
with open("C:/Users/nnh8/OneDrive - NIST/Documents/AM-CDM-main/OwlModels/AM-CDM-build.owl","r") as file:
      for ln in file:
            o_lines.append(ln.strip())
            i -= 1
            if i <= 0:
                break

cdd_lines=[] 
f="C:/Users/nnh8/OneDrive - NIST/Documents/ASTM-F3490-21-2.pdf"
with open(f,"rb") as file:
    pdf_reader = PyPDF2.PdfReader(file)
    for page_num in range(len(pdf_reader.pages)):
        page = pdf_reader.pages[page_num]
        text = page.extract_text()
        cdd_lines.extend(text.split('\n'))


def compare_fl(olst,cddlst):
    count_o=0
    newlst=[]
    for count_o,i in enumerate(olst):
        if(i not in cddlst):
            newlst.append((count_o,i))
    return newlst



print(compare_fl(o_lines,cdd_lines))
    
```
