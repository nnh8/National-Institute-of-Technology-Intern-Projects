```python
import tabula
import pandas as pd

def extract_and_save_table(pdf_path, page_number, csv_path):
    # Extract tables from the specified page of the PDF
    tables = tabula.read_pdf(pdf_path, pages=page_number, multiple_tables=True)

    # Assume the first table on the page is the one you want
    if tables:
        selected_table = tables[0]

        # Save the table to a CSV file
        selected_table.to_csv(csv_path, index=False)
        print(f"Table saved to: {csv_path}")
    else:
        print("No tables found on the specified page.")

# Specify the path to your PDF file
pdf_file_path = "try.pdf"

# Specify the page number (1-indexed) containing the table
page_number = 1

# Specify the path for the output CSV file
csv_file_path = "output_table.csv"

# Call the function to extract and save the table
extract_and_save_table(pdf_file_path, page_number, csv_file_path)
```
