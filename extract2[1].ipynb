```python
import fitz  # PyMuPDF
import pandas as pd

def extract_and_save_table(pdf_path, page_number, csv_path):
    # Open the PDF file
    pdf_document = fitz.open(pdf_path)

    # Get the specified page
    page = pdf_document[page_number - 1]

    # Extract text from the page
    page_text = page.get_text()

    # Assume tables are separated by lines
    table_lines = [line for line in page_text.split('\n') if line.strip()]

    # Split each line into columns (assuming tab-delimited data)
    table_data = [line.split('\t') for line in table_lines]

    # Create a DataFrame
    table_df = pd.DataFrame(table_data)

    # Save the table to a CSV file
    table_df.to_csv(csv_path, index=False, header=False)
    print(f"Table saved to: {csv_path}")

# Specify the path to your PDF file
pdf_file_path = "ASTM-F3490-21-2.pdf"

# Specify the page number (1-indexed) containing the table
page_number = 1

# Specify the path for the output CSV file
csv_file_path = "output_table.csv"

# Call the function to extract and save the table
extract_and_save_table(pdf_file_path, page_number, csv_file_path)
```

```python
import pdfplumber
import pandas as pd

def extract_text_by_keyword(pdf_path, keyword):
    # Open the PDF file
    with pdfplumber.open(pdf_path) as pdf:
        extracted_data = []

        # Iterate through all pages in the PDF
        for page_number, page in enumerate(pdf.pages, start=1):
            # Extract text from the page
            page_text = page.extract_text()

            # Check if the keyword is present in the page text
            if keyword.lower() in page_text.lower():
                # Append the page number and relevant text to the extracted data
                extracted_data.append({'Page Number': page_number, 'Text': page_text})

    return extracted_data

# Specify the path to your PDF file
pdf_file_path = "ASTM-F3490-21-2.pdf"

# Specify the keyword you want to search for
search_keyword = "AM"

# Call the function to extract text containing the keyword
extracted_data = extract_text_by_keyword(pdf_file_path, search_keyword)

# Create a DataFrame from the extracted data
df = pd.DataFrame(extracted_data)

# Specify the path for the output Excel file
excel_file_path = "output_table.xlsx"

# Save the DataFrame to an Excel file
df.to_excel(excel_file_path, index=False, engine='openpyxl')

print(f"Data extracted and saved to: {excel_file_path}")
```
