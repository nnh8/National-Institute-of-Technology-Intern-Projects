```python
import pdfplumber
import pandas as pd
import os

def extract_text_by_keyword(pdf_path, keyword):
    # Open the PDF file
    with pdfplumber.open(pdf_path) as pdf:
        extracted_data = []

        # Iterate through all pages in the PDF
        for page_number, page in enumerate(pdf.pages, start=1):
            # Extract text from the page
            page_text = page.extract_text()

            # Check if the keyword is present in the page text
            if keyword.lower() in page_text.lower():
                # Append the page number and relevant text to the extracted data
                extracted_data.append({'Page Number': page_number, 'Text': page_text})

    return extracted_data

# Specify the path to your PDF file
pdf_file_path = "ASTM-F3490-21-2.pdf"

# Specify the keyword you want to search for
search_keyword = "Project"

# Call the function to extract text containing the keyword
extracted_data = extract_text_by_keyword(pdf_file_path, search_keyword)

# Create a DataFrame from the extracted data
df = pd.DataFrame(extracted_data)

# Specify the relative file path
relative_excel_file_path = "C:/Users/nnh8/OneDrive - NIST/Documents/j12.xlsx"


# Get the absolute file path
full_excel_file_path = os.path.abspath(relative_excel_file_path)

# Save the DataFrame to an Excel file with the specified encoding
df.to_excel(full_excel_file_path, index=False, engine='openpyxl')



print(f"Data extracted and saved to: {excel_file_path}")
```

```python
import fitz
import pandas as pd
import os

def extract_text_from_pdf(pdf_path, start_page=0, end_page=None):
    pdf_document = fitz.open(pdf_path)
    
    if end_page is None:
        end_page = pdf_document.page_count

    extracted_text = ''
    for page_num in range(start_page, end_page):
        page = pdf_document[page_num]
        extracted_text += page.get_text()

    return extracted_text.encode('utf-8', 'ignore').decode('utf-8')


def parse_and_save_to_excel(extracted_text, excel_path):
    lines = extracted_text.split('\n')
    
    data = {'Text1': [], 'Text2': []}
    current_text = ''
    
    for line in lines:
        if '\t' in line:
            text1, text2 = line.split('\t')
            data['Text1'].append(text1)
            data['Text2'].append(text2)
        else:
            current_text += line.strip()
            data['Text1'].append(current_text)
            data['Text2'].append('')
            current_text = ''

    df = pd.DataFrame(data)
    df.to_excel(excel_path, index=False, engine='openpyxl')

    
pdf_file_path = "ASTM-F3490-21-2.pdf"
excel_file_path = os.path.abspath("C:/Users/nnh8/OneDrive - NIST/meg.xlsx")

start_page = 5
end_page = 17 # Adjusted the start and end page accordingly


extracted_text = extract_text_from_pdf(pdf_file_path, start_page, end_page)
parse_and_save_to_excel(extracted_text, excel_file_path)

print(f"Data extracted and saved to: {excel_file_path}")
```
